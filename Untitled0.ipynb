```python
!pip install yfinance -q
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib .pyplot as plt

#pick a stock - example: apple
ticker= "AAPL"
data= yf.download(ticker,start="2022-01-01",end="2024-01-01")

data.head()

# Plot closing price over time
plt.figure(figsize=(10,5))
plt.plot(data['Close'])
plt.title(f'{ticker} Closing Price')
plt.xlabel('Date')
plt.ylabel('Price (USD)')
plt.show()

# Daily returns = % change from one day to the next
data['Daily Return'] = data['Close'].pct_change()

# Plot histogram of daily returns
plt.figure(figsize=(8,5))
data['Daily Return'].hist(bins=50)
plt.title(f'{ticker} Daily Returns Distribution')
plt.xlabel('Daily Return')
plt.show()

# Basic stats
print("Mean daily return:", data['Daily Return'].mean())
print("Volatility (std dev):", data['Daily Return'].std())

data['MA20'] = data['Close'].rolling(window=20).mean()
data['MA50'] = data['Close'].rolling(window=50).mean()

plt.figure(figsize=(10,5))
plt.plot(data['Close'], label='Close Price', alpha=0.5)
plt.plot(data['MA20'], label='20-day MA')
plt.plot(data['MA50'], label='50-day MA')
plt.legend()
plt.title(f'{ticker} Price with Moving Averages')
plt.show()

# التقلب المتحرك لآخر 30 يوم (Rolling 30-day volatility)
data['Rolling Volatility'] = data['Daily Return'].rolling(window=30).std()

plt.figure(figsize=(10,5))
plt.plot(data['Rolling Volatility'])
plt.title(f'{ticker} 30-Day Rolling Volatility')
plt.xlabel('Date')
plt.ylabel('Volatility (Std Dev)')
plt.show()

from sklearn.linear_model import LinearRegression

# نجهز البيانات: X = رقم اليوم، y = السعر
data = data.dropna()  # نشيل أي صفوف فيها NaN (بسبب المتوسطات المتحركة)
data['Day Number'] = np.arange(len(data))

X = data[['Day Number']]
y = data['Close']

model = LinearRegression()
model.fit(X, y)

# التنبؤ
data['Predicted'] = model.predict(X)

plt.figure(figsize=(10,5))
plt.plot(data['Close'], label='Actual Price', alpha=0.6)
plt.plot(data['Predicted'], label='Linear Trend (Predicted)', color='red')
plt.legend()
plt.title(f'{ticker} Actual vs Linear Trend')
plt.show()

print("Model slope (daily price change trend):", model.coef_[0])
```

## Conclusion

This project explored basic statistical and time-series techniques applied to
AAPL stock data (2022–2024):

- **Price trend**: AAPL declined significantly through 2022, likely reflecting
  broader market pressure from rising interest rates, before recovering steadily
  through 2023.
- **Daily returns**: The distribution of daily returns is roughly bell-shaped but
  exhibits fat tails — extreme price movements occur more often than a normal
  distribution would predict, a well-known property of financial return data.
- **Moving averages**: The 20-day and 50-day moving averages smooth out daily
  noise and reveal the underlying trend. A "golden cross" (20-day MA crossing
  above the 50-day MA) coincided with the 2023 recovery.
- **Volatility**: Rolling 30-day volatility was highest during the 2022 downturn
  and declined steadily afterward — confirming that periods of sharp price
  decline tend to coincide with higher volatility.
- **Forecasting**: A simple linear regression captures the overall upward trend
  but fails to explain short-term fluctuations, illustrating the limitations of
  naive linear models for financial forecasting.

**Limitations**: This analysis is for educational purposes only. Stock prices
are influenced by countless factors not captured here (news, earnings,
macroeconomic events), and linear trend models should never be used as a
trading strategy.
